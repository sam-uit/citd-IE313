### Xử Lý Giá Trị Bị Khuyết

```python
# none NA df
pd.DataFrame(data = {
    'a': [1, 2, 3, 4, 5],
    'b': [6, 7, 8, 9, 10],
    'c': [11, 12, 13, 14, 15]
})
```

```python
import numpy as np
```

```python
# np.NA df
pd.DataFrame(data = {
    'a': [1, np.nan, 3, 4, 5],
    'b': [np.nan, 7, 8, 9, '?'],
    'c': [11, 12, 13, np.nan, 15]
})
```

#### Loại Bỏ Giá Trị Khuyết (`.dropna()`)

```python
# dropna sample
pd.DataFrame(data = {
    'a': [1, np.nan, 3, 4, 5],
    'b': [np.nan, 7, 8, 9, '?'],
    'c': [11, 12, 13, np.nan, 15]
}).dropna()
```

```python
# dropna sample
pd.DataFrame(data = {
    'a': [1, np.nan, 3, 4, 5],
    'b': [np.nan, 7, 8, 9, '?'],
    'c': [11, 12, 13, np.nan, 15]
}).dropna(subset=['a'])
```

#### Thay Thế Giá Trị Khuyết (`.replace()`)

```python
# replace sample
df2 = pd.DataFrame(data = {
    'a': [1, np.nan, 3, 4, 5],
    'b': [np.nan, 7, 8, 9, '?'],
    'c': [11, 12, 13, np.nan, 15]
})
```

```python
df2['c'].replace(to_replace=np.nan, value=df2['c'].mean())
```

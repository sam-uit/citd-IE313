## 6. Chọn ra các cột số, chuẩn hóa thành `zscore`.

### Nhận Xét

- `df` có một số lượng lớn các cột có dạng số (numeric), bao gồm `int` và `float`.

````python
df.info()
````

````python
df.dtypes
````

### Kế Hoạch

- Lọc ra những cột có `dtype` là kiểu số.
- Sử dụng `sklearn.preprocessing.StandardScaler` để `transform` từng cột tương ứng.

````python
# Chọn ra những cột có dạng số
df.select_dtypes(include=['int64', 'float64']).info()
````

````python
# Lấy ra danh sách các cột này
columns_list = df.select_dtypes(include=['int64', 'float64']).columns.to_list()
columns_list
````

````python
# import StandardScaler
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
````

````python
# Thực hiện transform zscore
for column in columns_list:
    df[column + '_zscore'] = ss.fit_transform(df[[column]])
````

````python
# Kiểm lại df
df.shape
````

````python
# Xem sample
df.sample()
````

````python
# Lấy ra các cột zscore
df[[ column + '_zscore' for column in columns_list]]
````

### Kết Quả

- Có tổng số 16 cột dạng số.
- Tất cả các cột dạng số đã được tạo cột zscore, tương ứng 16 cột thêm mới.
- Kích thước của `df` hiện tại là (193, 42), với 42 = 26 cột đã có + 16 cột mới.

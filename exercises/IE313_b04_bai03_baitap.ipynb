## Bài Làm

### Đổi Tên Cột

```python
# khai báo danh sách cột đổi cũ: mới
columns = {
    'OdName': 'Country',
    'AreaName': 'Continent',
    'RegName': 'Region'
}

# kiểm thử sau khi đổi
df_canada.rename(columns=columns, inplace=False).info()
```

```python
# thực hiện và kiểm lại
df_canada.rename(columns=columns, inplace=True)
df_canada.info()
```

### Loại Bỏ Cột

```python
# danh sách các cột cần bỏ
drop_collumns = ['AREA', 'REG', 'DEV', 'Type', 'Coverage']

# kiểm thử
df_canada.drop(columns=drop_collumns, inplace=False).info()
```

```python
# thực hiện
df_canada.drop(columns=drop_collumns, inplace=True)
df_canada.info()
```

```python
# liệt kê tên các nước để đảm bảo gọi đúng tên
# ví dụ Viet Nam thay vì Vietnam, hay vietnam, vv..
df_canada['Country'].unique()
```

### Tạo Hàm Chọn

```python
# Đưa vào một df và tên quốc gia cụ thể.
# Trả về số lượng di cư của quốc gia đó tới Canada
# Các cột từ 1980 - 2013
# Mỗi cột là một năm
# Thêm cột tổng số ở cuối

def chon(df: pd.DataFrame, Country: str) -> pd.Series:
    # chọn ra quốc gia và tạo bản sao
    df_chon = df.loc[df['Country'] == Country].copy()
    # lấy các cột năm (các cột có name là số)
    years = [col_name for col_name in df.columns.to_list() if isinstance(col_name, int)]
    # tính tổng từ các năm và gán cột total
    df_chon['total'] = df_chon[years].sum(axis=1)
    # trả về series
    return df_chon
```

```python
# Kiểm thử

chon(df_canada, 'Viet Nam')
```

### Hoàn Thành

- Hàm `chon()` đã được hoàn thành.
- Nhận vào `pd.DataFrame` và một tên nước.
- Trả về một `pd.Series` chứa dòng của nước đó, kèm một cột tổng.

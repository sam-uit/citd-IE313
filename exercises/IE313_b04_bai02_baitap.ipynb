### Integer encoding cho make

Ý tưởng: lấy toàn bộ các giá trị trong `make`, sau đó map/gán tương ứng các giá trị số thành một `dict`, sau đó dùng `.replace()` để thực hiện việc chuyển đổi.

````python
# xem thuộc tính của make, hiện tại dtype là object
df['make'].info()
````

````python
# Các maker trong dataset
df['make'].unique()
````

````python
# mỗi maker được map vào một số tương ứng
makers_count = 0
makers = {}

for maker in df['make'].unique():
    makers_count += 1
    makers[maker] = makers_count

makers
````

````python
# thử trước khi replace
df['make'].replace(makers, inplace=False)
````

````python
# thực hiện replace
df['make'].replace(makers, inplace=True)
````

````python
# kiểm tra lại
df['make']
````

````python
# kiểm tra lại các loại
df['make'].unique()
````

````python
# dùng replace để cập nhật/chuyển đổi
# phần này giống câu ở trên
# vẫn nên kiểm thử trước
df['num-of-doors'].replace(doors, inplace=False).unique()
````

### One-hot encoding áp dụng còn lại.

````python
# xem nhanh df hiện tại
df.sample()
````

````python
# xác định các cột có dtype là object và chưa xử lý
df.info()
````

````python
# lấy danh sách các column và dtype tương ứng
data_types = df.dtypes
data_types.info()
````

````python
# lọc ra các column có dtype là object
# đây chính là các column còn lại cần chuyển đổi
object_columns = data_types[data_types == 'object'].index.tolist()
object_columns
````

````python
# thử lấy dummies của một column trong danh sách trên
# đảm bảo rằng dummies hợp lý để trả về true/false
pd.get_dummies(df[object_columns[0]], prefix=object_columns[0])
````

````python
# xác định thử các giá trị của một column trong danh sách
df[object_columns[0]].value_counts()
````

````python
# với mỗi column trong danh sách, áp dụng one-hot encoding
for cot in object_columns:
    dummies = pd.get_dummies(df[cot], prefix=cot)
    df = pd.concat([df, dummies], axis=1)

# sau đó kiểm tra lại số cột/dtype tương ứng
df.info()
````

````python
# để làm sạch, chúng ta sẽ xóa bỏ các column đã được chuyển đổi
# nói cách khác, là xóa các columns có dtype object
df = df.drop(columns=object_columns)

# sau đó kiểm lại
df.info()
````

## Chuẩn Hóa z-score (cho các cột từ bước 3)

### Nhận Xét

- Một số lượng lớn các cột mới đã được tạo ra từ bước chuyển đổi ở trên.
- Kèm theo các cột từ bài tập 03.

### Kế Hoạch

Có các cách sau:

1. Xóa tất cả các cột z-score đã tạo ở buổi 03, và tạo mới zscore cho tất cả các cột.
2. Lọc ra các cột mới từ bước 03 và tạo cột zscore tương ứng.

Trong bài này sẽ làm theo cách 2, vì sát với đề bài của bài tập.

````python
# xem lại các cột đã được tạo z-score ở bài của buổi 03
columns_list
````

````python
# đây là các zscore columns đã được tạo từ bài tập buổi 03
zscore_columns = [column + '_zscore' for column in columns_list]
zscore_columns
````

````python
# chọn ra các columns dạng số và bool
df.select_dtypes(include=['int64', 'float64', 'bool']).columns.to_list()
````

````python
# lấy ra những cột từ danh sách trên mà:
# 1. không thuộc các cột đã được tính zscore
# 2. không thuộc các cột là zscore
columns_to_zscore = [column for column in df.select_dtypes(include=['int64', 'float64', 'bool']).columns.to_list() if (column not in zscore_columns and column not in columns_list)]
columns_to_zscore
````

````python
df.shape
````

````python
# thực hiện tạo zscore cho các cột mới
# đồng thời thêm các cột zstore mới tạo vào danh sách các cột zstore trước đó

for column in columns_to_zscore:
    df[column + '_zscore'] = ss.fit_transform(df[[column]])
    zscore_columns.append(column + '_zscore')
````

- Shape hiện tại là (193, 100), trông đó 193 dòng đều non-null.
